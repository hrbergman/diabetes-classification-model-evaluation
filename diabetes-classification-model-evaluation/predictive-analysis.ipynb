## B1. Predictive Method Technique & Expected Outcomes


#### How A Decision Tree Analyzes the Dataset for Diabetes Prediction

A decision tree works to analyze the dataset by building a hierarchical structure of questions, each with a basic binary (Yes/No) answer. Starting at the root node, the tree evaluates a specific feature and directs data points down corresponding branches. This branching continues, with each internal node posing a question based on a different feature, until a terminal node, or 'leaf,' is reached. The leaf node assigns a final classification, such as 'diabetes: Yes' or 'diabetes: No.'

Unlike linear models, a decision tree can capture more complex, non-linear relationships. This makes them particularly useful for datasets like ours, where diabetes risk factors interact in intricate ways. Compared to K-Nearest Neighbors (KNN), a decision tree requires no feature scaling, is resilient to outliers, and handles imbalanced data more effectively (Afeworki, 2021). 

The tree's construction is driven by a process of feature selection and recursive splitting. At each node, the algorithm identifies the feature that best separates the data into groups with similar diabetes outcomes. For example, the initial split might be based on a key predictor like complication risk. Subsequent splits could refine these groups by considering age, BMI, and comorbidities like stroke or back pain. This iterative splitting continues until the tree accurately classifies the data, culminating in the leaf nodes that provide the final predictions.

#### Expected Outcomes of the Analysis
Our analysis will evaluate the model's performance using metrics such as accuracy, MSE, ROC and a confusion matrix. High performance across these metrics will indicate that our model effectively distinguishes between patients with and without diabetes.

<a id="b2"></a>
## B2. Assumption of the Chosen Prediction Method

Decision tree algorithms operate on the assumption that complex decisions can be broken down into a series of simpler, binary choices. At each node decision trees employ binary splitting, dividing data into two groups based on a single feature - essentially asking a simple yes-or-no question (Anshul, 2025).

## D2. Analysis Technique

### Step 1. Train/Test Split with Stratification

Before training our model, we need to split our dataset into training and testing subsets (which we performed previously in [Section D1](#d1)) to accurately assess performance and avoid overfitting. Given the class imbalance in our target variable (diabetes), with approximately 27% positive cases, we utilized stratified sampling. This technique helps guarantee that the proportion of diabetes cases is consistent across both training and testing sets, enabling a more reliable evaluation of the model's predictive capabilities.

### Step 2. Train a Base Decision Tree Using GridSearchCV

Decision trees can overfit easily if left untuned, so we’ll use GridSearchCV to help optimize parameters (tree depth, minimum samples per leaf) using cross-validation. This ensures our model is the best generalizable version, not just one that performs well on our training data. It helps prevent issues like overly complex trees (which memorize data) and oversimplified trees (which can miss meaningful patterns in our data). This provides us with a better balanced, well-tuned baseline model that will serve as our reference point, and also gives us some insight into how sensitive performance is to tree complexity. 

```python
# Decision Tree with GridSearchCV

# Define the base Decision Tree model
dtree = DecisionTreeClassifier(random_state=42)

# Define a parameter grid for tuning
param_grid = {
    'max_depth': [3, 5, 10, None],
    'min_samples_leaf': [1, 2, 4]
}

# Setup GridSearchCV
grid_search = GridSearchCV(
    estimator=dtree,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

# Fit GridSearchCV on training data
grid_search.fit(X_train, y_train)

# Best Decision Tree model
best_tree = grid_search.best_estimator_
print("Best Decision Tree Parameters:", grid_search.best_params_)
```

### Step 3. Evaluate Decision Tree Performance

Before applying boosting techniques, we should thoroughly evaluate the baseline decision tree's performance. A confusion matrix allows us to analyze the distribution of false positives and false negatives, highlighting potential areas for improvement. We supplement this with a classification report, which provides precision, recall, and F1-score, and calculate MSE/RMSE and AUC to assess model fit and class separation capabilities. This evaluation provides a good understanding of the baseline model's strengths and weaknesses.

```python
# Evaluate Decision Tree Predictions

# Predictions and Probabilities from the tuned Decision Tree
y_pred_tree = best_tree.predict(X_test)
y_proba_tree = best_tree.predict_proba(X_test)[:, 1]

# Evaluation Metrics

# Classification Report
print("Classification Report (Decision Tree):\n", classification_report(y_test, y_pred_tree))

# Confusion Matrix
print("Confusion Matrix (Decision Tree):")
sns.heatmap(confusion_matrix(y_test, y_pred_tree), annot=True, fmt='d', cmap='Blues')
plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# MSE and RMSE
mse_tree = mean_squared_error(y_test, y_pred_tree)
rmse_tree = np.sqrt(mse_tree)
print(f"Decision Tree MSE: {mse_tree:.4f}")
print(f"Decision Tree RMSE: {rmse_tree:.4f}")

# AUC Score
auc_tree = roc_auc_score(y_test, y_proba_tree)
print(f"Decision Tree AUC Score: {auc_tree:.4f}")
```

Through GridSearchCV, we determined the optimal decision tree configuration: max_depth=3 and min_samples_leaf=1. This configuration resulted in an AUC of 0.5204. Though this means our model exceeds random classification, it only does so by a marginal amount. This result suggests the decision tree is a weak learner, struggling to capture complex patterns. To address this, we’ll apply AdaBoost.

### Step 4. Boost the Decision Tree with AdaBoost

To enhance predictive performance, we use AdaBoost, an ensemble technique that combines multiple weak learners (the GridSearchCV-optimized decision tree) into a strong classifier. AdaBoost adaptively adjusts the weights of training instances, emphasizing those misclassified by previous learners. This sequential learning process results in a model with improved accuracy, precision, and recall, while also reducing the risk of overfitting associated with deep, single decision trees. Using our tuned decision tree as the base learner ensures a strong foundation for boosting.

```python
# Build AdaBoost with tuned decision tree as base estimator

# AdaBoost using the best decision tree
ada_model = AdaBoostClassifier(
    base_estimator=best_tree,
    n_estimators=100,
    learning_rate=0.5,
    random_state=42
)

# Train AdaBoost model
ada_model.fit(X_train, y_train)
```

### Step 5. Evaluate AdaBoost Predictions

To determine the extent to which the AdaBoost model enhances predictive performance, we’ll compare it to the baseline decision tree. This analysis allows us to quantify the impact of boosting on metrics such as accuracy, precision, recall, and AUC.

```python
# Evaluate final AdaBoost model

# Predictions and Probabilities
y_pred_ada = ada_model.predict(X_test)
y_proba_ada = ada_model.predict_proba(X_test)[:, 1]

# Evaluation Metrics

# Classification Report
print("Classification Report (AdaBoost):\n", classification_report(y_test, y_pred_ada))

# Confusion Matrix
print("Confusion Matrix (AdaBoost):")
sns.heatmap(confusion_matrix(y_test, y_pred_ada), annot=True, fmt='d', cmap='Greens')
plt.title("AdaBoost Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# MSE and RMSE
mse_ada = mean_squared_error(y_test, y_pred_ada)
rmse_ada = np.sqrt(mse_ada)
print(f"AdaBoost MSE: {mse_ada:.4f}")
print(f"AdaBoost RMSE: {rmse_ada:.4f}")

# AUC Score
auc_ada = roc_auc_score(y_test, y_proba_ada)
print(f"AdaBoost AUC Score: {auc_ada:.4f}")
```

## E4. Recommended Course of Action

Given the disappointing performance of the boosted model, particularly its lower accuracy and higher propensity to error compared to the initial model, it needs much more refinement to ensure reliability in real-world application to effectively identify patients with diabetes.

Throughout this analysis and our previous projects, we’ve discovered our models tend to exhibit a significant bias towards classifying patients as non-diabetic, resulting in zero true and false positives. This bias is likely due to a class imbalance issue, stemming from the overrepresentation of non-diabetic patients in our dataset. 

To mitigate this, we can implement resampling techniques such as oversampling (increasing the number of diabetic patients we have in our training data) or undersampling (reducing the number of non-diabetic patients in our training data). We can also adjust class weights to penalize misclassifications of diabetic cases, to help encourage our model to focus on detecting diabetic patients.

Additionally, we should consider evaluating our model using other metrics that are more suitable for imbalanced data, such as the Precision-Recall Curve and F1 Score (rather than Accuracy and MSE). These metrics focus on how well the model identifies diabetic cases rather than relying solely on accuracy, providing a more comprehensive assessment of performance.
